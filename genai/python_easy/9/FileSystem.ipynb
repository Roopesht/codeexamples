### Read entire file

```python
file = open('sample.txt', 'r')
content = file.read()
print("Full content:", content)
print ("Type of content: ", type(content))
file.close()

content = file.read()
```

### How to read line by line

```python
file = open('sample.txt', 'r')
first_line = file.readline()
second_line = file.readline()
print("First line:", first_line.strip())
print("Second line:", second_line.strip())
file.close()
```

### Readlines

```python
file = open('sample.txt', 'r')
all_lines = file.readlines()
print("All lines:", all_lines)
print ("Type of all_lines", type(all_lines))
file.close()
```

### Iteration

```python
file = open('sample.txt', 'r')
while True:
    line = file.readline()
    if not line:
        break
    print(line.strip())
file.close()
```

```python

lines = ["1","2","3","4"]
lines1 = [line + "\n" for line in lines ]
file = open('multiple_lines.txt', 'w')
file.writelines(lines1)
file.close()
```

### Context manager

```python
# original
file = open('sample.txt', 'r')
content = file.read()
print("Full content:", content)
file.close()
```

```python
with open('sample.txt', 'r') as file:
  content = file.read()
  print("Full content:", content)
```
