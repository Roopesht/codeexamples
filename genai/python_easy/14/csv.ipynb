```python
import csv

# Reading CSV file
with open('a.csv', 'r') as file:
    reader = csv.reader(file)
    headers = next(reader)  # Get headers
    print (headers)
    for row in reader:
        print(row)
```

```python
with open('a.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row['id'], row[' name'])
```

```python
with open('a.csv', 'w') as file:
  writer = csv.writer(file)
  writer.writerow(['id', 'name'])
  writer.writerow(['1', 'Mohan'])
  writer.writerow(['2', 'Vishwa'])
  writer.writerow(['3', 'Rohit'])
  writer.writerow(['4', 'Kiran'])


```

```python
with open('a.csv', 'a') as file:
  writer = csv.writer(file)
  writer.writerow(['Jalaja', '10'])
```

```python
with open('a.csv', 'a') as file:
  fieldnames = ['id', 'name']
  writer = csv.DictWriter(file, fieldnames)
  writer.writerow({'name': 'Kolappa', 'id': '10'})
```
