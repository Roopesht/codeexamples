# Concept 2: Cost and Latency Vary by Model

Runs the exact same prompt against a small and a large model, timing and pricing both.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
# Prices are illustrative ($ per 1M tokens) - check the provider's current pricing page for real numbers
PRICING = {"gpt-4o-mini": {"input": 0.15, "output": 0.60}, "gpt-4o": {"input": 2.50, "output": 10.00}}
```

```python
import time
messages = [{"role": "user", "content": "Explain what an API is, in 3 sentences, for a complete beginner."}]
for model in ["gpt-4o-mini", "gpt-4o"]:
    start = time.time()
    response = client.chat.completions.create(model=model, messages=messages)
    elapsed = time.time() - start
    usage = response.usage
    cost = (usage.prompt_tokens * PRICING[model]["input"] + usage.completion_tokens * PRICING[model]["output"]) / 1_000_000
    print(model, f"{elapsed:.2f}s", f"${cost:.6f}", usage.total_tokens, "tokens")
```

## How to Extend
Run this cell twice and compare latency both times — notice how much it varies run to run. Try a task where the larger model's answer is visibly better, not just slower.
