# Concept 3: A Unified Way to Compare

Runs the same two-model comparison through LiteLLM's single interface instead of juggling per-provider SDKs.

## Setup
Store your key in Colab **Secrets** as `OPENAI_API_KEY`.

```python
import os
from google.colab import userdata
os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
```

```python
from litellm import completion
messages = [{"role": "user", "content": "Explain what an API is, in 3 sentences, for a complete beginner."}]
for model in ["gpt-4o-mini", "gpt-4o"]:
    response = completion(model=model, messages=messages)
    print(model, response.usage)
```

## How to Extend
Swap one model string for a different provider's model (e.g. an Anthropic or Ollama model, per LiteLLM's docs) and see the same code shape still work.
