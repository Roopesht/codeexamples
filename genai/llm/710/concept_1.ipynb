# Concept 1: Demo vs. Product

Makes one real call and turns its token usage into an actual dollar cost.

## Setup
Store your key in Colab **Secrets** (key icon, left sidebar) as `OPENAI_API_KEY`.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
# Prices are illustrative ($ per 1M tokens) - check the provider's current pricing page for real numbers
PRICING = {"gpt-4o-mini": {"input": 0.15, "output": 0.60}, "gpt-4o": {"input": 2.50, "output": 10.00}}
```

```python
messages = [{"role": "user", "content": "Explain what an API is, in 3 sentences, for a complete beginner."}]
response = client.chat.completions.create(model="gpt-4o-mini", messages=messages)
usage = response.usage
cost = (usage.prompt_tokens * PRICING["gpt-4o-mini"]["input"] + usage.completion_tokens * PRICING["gpt-4o-mini"]["output"]) / 1_000_000
print(usage.prompt_tokens, usage.completion_tokens, f"${cost:.6f}")
```

## How to Extend
Multiply this cost by 10,000 (a realistic daily call volume) and see what it becomes. Try a longer prompt and watch the cost scale with prompt_tokens.
