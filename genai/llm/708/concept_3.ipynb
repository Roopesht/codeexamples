# Concept 3: Running an Eval Suite

Writes a small Promptfoo config with 3 test cases and runs them all at once from the command line.

## Setup
Store your key in Colab **Secrets** as `OPENAI_API_KEY`. Promptfoo (a Node CLI) reads it from the shell environment, not from a Python client object.

```python
import os
from google.colab import userdata
os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
```

```python
config = '''
prompts:
  - "Extract the order number from this message. Respond with only the order number.\\n\\nMessage: {{message}}"
providers:
  - openai:gpt-4o-mini
tests:
  - vars: {message: "Hi, I'd like to check on order #A1234, it hasn't arrived yet."}
    assert: [{type: contains, value: "A1234"}]
  - vars: {message: "My order number is 98765 and I want a refund."}
    assert: [{type: contains, value: "98765"}]
  - vars: {message: "Please cancel order 4432 immediately!"}
    assert: [{type: contains, value: "4432"}]
'''
with open("promptfooconfig.yaml", "w") as f:
    f.write(config)
```

```python
!npx --yes promptfoo eval -c promptfooconfig.yaml
```

## How to Extend
Add a 4th test case with a message that has no order number at all, asserting the response does NOT contain a made-up number. Change one `assert value` to something wrong and watch it fail.
