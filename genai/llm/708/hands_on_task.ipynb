# Hands-on Task: 5 Test Cases

Writes 5 varied test cases (including one edge case) for the order-extraction prompt and runs them all through Promptfoo.

## Setup
Store your key in Colab **Secrets** as `OPENAI_API_KEY`.

```python
import os
from google.colab import userdata
os.environ["OPENAI_API_KEY"] = userdata.get("OPENAI_API_KEY")
```

```python
config = '''
prompts:
  - "Extract the order number from this message. Respond with only the order number, or NONE if there isn't one.\\n\\nMessage: {{message}}"
providers:
  - openai:gpt-4o-mini
tests:
  - vars: {message: "Hi, I'd like to check on order #A1234, it hasn't arrived yet."}
    assert: [{type: contains, value: "A1234"}]
  - vars: {message: "My order number is 98765 and I want a refund."}
    assert: [{type: contains, value: "98765"}]
  - vars: {message: "Order: ORD-556-XZ was delivered to the wrong address."}
    assert: [{type: contains, value: "ORD-556-XZ"}]
  - vars: {message: "I haven't placed any orders yet, just browsing."}
    assert: [{type: contains, value: "NONE"}]
  - vars: {message: "Please cancel order 4432 immediately!"}
    assert: [{type: contains, value: "4432"}]
'''
with open("promptfooconfig.yaml", "w") as f:
    f.write(config)
```

```python
!npx --yes promptfoo eval -c promptfooconfig.yaml
```

## How to Extend
For any case that fails, print `extract(message)` directly in Python to see the raw output and understand why. Try tightening the prompt until all 5 pass.
