# Concept 1: "It Looked Fine" Isn't Testing

Runs the same extraction prompt against two real messages and shows one working cleanly while the other doesn't.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
def extract(message):
    prompt = f"Extract the order number from this message. Respond with only the order number.\n\nMessage: {message}"
    response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}])
    return response.choices[0].message.content
```

```python
messages = [
    "Hi, I'd like to check on order #A1234, it hasn't arrived yet.",
    "Please cancel order 4432 immediately!",
]
for msg in messages:
    print(msg, "->", extract(msg))
```
