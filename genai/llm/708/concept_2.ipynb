# Concept 2: Writing a Test Case

Turns 'does this look right' into a real, automatic pass/fail check using a simple assertion.

## Setup
Store your key in Colab **Secrets** (key icon, left sidebar) as `OPENAI_API_KEY`.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
def extract(message):
    prompt = f"Extract the order number from this message. Respond with only the order number.\n\nMessage: {message}"
    response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}])
    return response.choices[0].message.content
```

```python
message = "Hi, I'd like to check on order #A1234, it hasn't arrived yet."
response = extract(message)
passed = "A1234" in response
print(response, "-> PASS" if passed else "-> FAIL")
```

## How to Extend
Write a second assertion for a different message and expected order number. Try a stricter check (`response.strip() == "A1234"`) and see if the looser prompt still passes it.
