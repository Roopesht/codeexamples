# Concept 2: JSON Schema — The Contract

Defines a schema for a weather lookup, then checks a good and a bad response against it.

```python
!pip install -q jsonschema
```

```python
from jsonschema import validate, ValidationError
schema = {
    "type": "object",
    "properties": {
        "city": {"type": "string"},
        "unit": {"type": "string", "enum": ["celsius", "fahrenheit"]}
    },
    "required": ["city", "unit"]
}
good = {"city": "Pune", "unit": "celsius"}
bad = {"city": "Pune"}
for candidate in (good, bad):
    try:
        validate(instance=candidate, schema=schema)
        print(candidate, "-> VALID")
    except ValidationError as e:
        print(candidate, "-> INVALID:", e.message)
```

## How to Extend
Add a third candidate with `unit` set to `"kelvin"` and see the enum catch it. Add a new required field and watch the existing candidates start failing until you update them.
