# Concept 2: JSON Schema — The Contract

A JSON Schema is a precise, machine-checkable description of what a valid response looks like. It's documentation for a human and a validation rule for your code, in one place.

## What a Schema Defines
- **Types:** string, number, boolean, array, object
- **Required fields:** which keys must be present
- **Enums:** a fixed set of allowed values
- **Descriptions:** plain-language hints the model reads too

## A Schema for `get_weather`

```python
get_weather_schema = {

    "type": "object",

    "properties": {

        "city": {"type": "string"},

        "unit": {"type": "string", "enum": ["celsius", "fahrenheit"]}

    },

    "required": ["city", "unit"]

}

get_weather_schema
```

## Demo: Validating Responses Against the Schema
One response is valid, one is missing a required field. Watch validation catch it.

```python
from jsonschema import validate, ValidationError



good = {"city": "Pune", "unit": "celsius"}

bad = {"city": "Pune"}  # missing "unit"



for candidate in (good, bad):

    try:

        validate(instance=candidate, schema=get_weather_schema)

        print(candidate, "-> VALID")

    except ValidationError as e:

        print(candidate, "-> INVALID:", e.message)
```
