# Concept 1: Why Free Text Isn't Enough

The model isn't the problem — parsing its prose is. This notebook shows how a naive regex-based parser breaks the moment an LLM response is phrased slightly differently, even though both responses mean exactly the same thing.

## The Problem
- LLMs default to prose: fluent sentences, not data
- Programs need predictability: the same field, in the same shape, every time
- 'Reasonable-sounding' text is not the same as 'parseable' text

## Demo: Parsing Free-Text LLM Output
Two responses below mean the same thing — 24°C and sunny in Pune — but are worded differently, the way a real LLM's output varies between calls.

```python
import re



# Simulated LLM responses to the same question, worded differently

responses = [

    "It's sunny today, around 24 degrees Celsius.",

    "Pune is looking sunny, roughly 24C right now.",

]



# Naive regex assuming a fixed sentence shape

pattern = r"around (\d+) degrees"



for text in responses:

    match = re.search(pattern, text)

    result = match.group(1) if match else "PARSE FAILED"

    print(f"{text!r} -> {result}")
```

## What Just Happened
The regex matched the first response but failed on the second, even though both carry the exact same information. This is the core reliability problem structured output solves.
