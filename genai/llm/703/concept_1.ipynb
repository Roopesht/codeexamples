# Concept 1: Why Free Text Isn't Enough

Asks the model the same question twice and tries to pull the temperature out with a simple pattern match.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
import re
prompt = "Invent a plausible current temperature and condition for Pune, one sentence, no disclaimers."
responses = []
for _ in range(2):
    reply = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}], temperature=1.0)
    responses.append(reply.choices[0].message.content)
pattern = r"around (\d+) degrees"
for text in responses:
    match = re.search(pattern, text)
    print(text, "->", match.group(1) if match else "PARSE FAILED")
```

## How to Extend
Call the model a third time and see the same regex fail again. Try a different city. Try loosening the pattern to catch more phrasings, and notice how quickly that becomes unmanageable.
