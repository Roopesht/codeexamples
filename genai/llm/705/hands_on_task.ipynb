# Hands-on Task: Embed, Store, Search

Embeds 20 short text snippets, stores them in a Chroma collection, and runs a query that should return the right match.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

## Your Snippets
Ships with the sample 20-snippet bank below. Replace `snippets` with your own list of 20 short text snippets to use your own content.

```python
import urllib.request
url = "https://raw.githubusercontent.com/Roopesht/codeexamples/main/genai/llm/705/docs/snippets.txt"
snippets = urllib.request.urlopen(url).read().decode().strip().split("\n")
```

```python
import chromadb
from chromadb.utils import embedding_functions
openai_ef = embedding_functions.OpenAIEmbeddingFunction(api_key=userdata.get("OPENAI_API_KEY"), model_name="text-embedding-3-small")
db = chromadb.Client()
collection = db.create_collection("snippets", embedding_function=openai_ef)
collection.add(documents=snippets, ids=[str(i) for i in range(len(snippets))])
```

```python
query = "What are baby dogs called?"
result = collection.query(query_texts=[query], n_results=1)
print("Top match:", result["documents"][0][0])
```

## How to Extend
Try `n_results=3` and check the next two matches make sense too. Swap in your own 20 snippets and a query only one of them can answer.

## Done When
The printed top match is the snippet the query was actually about.
