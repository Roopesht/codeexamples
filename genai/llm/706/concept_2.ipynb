# Concept 2: The Decide-or-Answer Branch

Gives the model two real tools and the same compound question, and watches it decide which to call.

## Setup
Store your key in Colab **Secrets** (key icon, left sidebar) as `OPENAI_API_KEY`.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
tools = [
    {"type": "function", "function": {"name": "get_date", "description": "Get today's date", "parameters": {"type": "object", "properties": {}}}},
    {"type": "function", "function": {"name": "calculate_tip", "description": "Calculate a tip amount", "parameters": {"type": "object", "properties": {"bill": {"type": "number"}, "percent": {"type": "number"}}, "required": ["bill", "percent"]}}},
]
```

```python
prompt = "What's 15% tip on $42.50, and also what's today's date?"
response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}], tools=tools)
for call in response.choices[0].message.tool_calls:
    print(call.function.name, call.function.arguments)
```

## How to Extend
Ask something needing neither tool and confirm `tool_calls` comes back empty. Remove one tool and see how the model handles the half it can no longer answer.
