# Concept 2: The Decide-or-Answer Branch

Gives the model one real tool for the same question, and watches it decide to call it.

## Setup
Store your key in Colab **Secrets** (key icon, left sidebar) as `OPENAI_API_KEY`.

```python
!pip install -q openai
```

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
tools = [
    {"type": "function", "function": {"name": "get_next_batch_date", "description": "Get the start date of the next AI Engineer course batch", "parameters": {"type": "object", "properties": {}}}},
]
```

```python
prompt = "When's the next batch of the AI Engineer course starting?"
response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}], tools=tools)
call = response.choices[0].message.tool_calls[0]
print(call.function.name, call.function.arguments)
```

## How to Extend
Ask something unrelated to batches and confirm tool_calls comes back empty.
