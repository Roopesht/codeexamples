# Concept 5: Real-World Actions & Handing Off

Actually sends a live push notification via ntfy.sh (free, no signup) instead of just printing a mocked result - a real hand-off to a human.

## Setup
No API key needed for this one - just a unique topic name below, so only you see your own notifications (ntfy.sh topics are public by name).

```python
!pip install -q requests
```

```python
NTFY_TOPIC = "ojasa-mirai-agent-demo-CHANGE-ME"
import requests
def notify_admissions(message):
    r = requests.post(f"https://ntfy.sh/{NTFY_TOPIC}", data=message.encode("utf-8"))
    return f"notified admissions (status {r.status_code})"
```

```python
print("Open https://ntfy.sh/" + NTFY_TOPIC + " in another tab, then run this cell.")
print(notify_admissions("A student's friend wants to join the next AI Engineer batch - please follow up."))
```

## How to Extend
Subscribe to your topic in the ntfy phone app (or the web page) to see it arrive as a real push notification, not just a browser tab.
