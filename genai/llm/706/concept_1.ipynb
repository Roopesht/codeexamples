# Concept 1: Agent vs. Single Call

Asks a real model a compound question with no tools available, and watches it guess the part it can't actually know.

## Setup
Store your key in Colab **Secrets** (key icon, left sidebar) as `OPENAI_API_KEY`.

```python
!pip install -q openai
```

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
prompt = "What's 15% tip on $42.50, and also what's today's date?"
response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}])
print(response.choices[0].message.content)
```

## How to Extend
Ask a question needing only knowledge it has (no date/math) and see it succeed cleanly — the limitation is specific, not general.
