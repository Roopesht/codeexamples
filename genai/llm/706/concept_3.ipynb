# Concept 3: Explaining Its Own Decision

Adds a required 'reason' field to each tool's schema, so the model states why it picked a tool alongside the call itself.

## Setup
Store your key in Colab **Secrets** (key icon, left sidebar) as `OPENAI_API_KEY`.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
tools = [
    {"type": "function", "function": {"name": "get_date", "description": "Get today's date", "parameters": {"type": "object", "properties": {"reason": {"type": "string"}}, "required": ["reason"]}}},
    {"type": "function", "function": {"name": "calculate_tip", "description": "Calculate a tip amount", "parameters": {"type": "object", "properties": {"bill": {"type": "number"}, "percent": {"type": "number"}, "reason": {"type": "string"}}, "required": ["bill", "percent", "reason"]}}},
]
```

```python
import json
prompt = "What's 15% tip on $42.50, and also what's today's date?"
response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}], tools=tools)
for call in response.choices[0].message.tool_calls:
    args = json.loads(call.function.arguments)
    print(call.function.name, "->", args["reason"])
```

## How to Extend
Ask an ambiguous question and read the reason to see how the model is interpreting it. Try a prompt where the 'right' tool is debatable and see which way it leans, and why.
