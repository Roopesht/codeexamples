Asks a real model a question only a private company's docs could answer, with no context given.

## Setup
Store your key in Colab **Secrets** (key icon, left sidebar) as `OPENAI_API_KEY`.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
prompt = "What's Ojasa Mirai's refund policy for its courses?"
response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}])
print(response.choices[0].message.content)
```

## How to Extend
Try a made-up product or policy name of your own and see the same confident guessing. Ask it to say "I don't know" if unsure, and compare how often it actually does.
