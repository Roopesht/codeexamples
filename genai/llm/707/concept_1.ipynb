# Concept 1: When One Prompt Isn't Enough

Asks one prompt to both summarize and classify a review at once, and watches the output come back inconsistently shaped.

## Setup
Store your key in Colab **Secrets** (key icon, left sidebar) as `OPENAI_API_KEY`.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
review = "The battery life is incredible and the camera quality blew me away, though the phone does run a bit warm during long gaming sessions."
```

```python
prompt = f"Summarize this review in one sentence, then classify it as positive or negative:\n{review}"
response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": prompt}])
print(response.choices[0].message.content)
```

## How to Extend
Run this cell 3 times and compare the shape of the output each time. Try a stricter prompt ("respond as: SUMMARY: ... LABEL: ...") and see how far formatting alone gets you.
