# Concept 2: Chaining Calls

Splits the same task into two focused calls, feeding the first call's real output into the second.

## Setup
Store your key in Colab **Secrets** (key icon, left sidebar) as `OPENAI_API_KEY`.

```python
from openai import OpenAI
from google.colab import userdata
client = OpenAI(api_key=userdata.get("OPENAI_API_KEY"))
```

```python
review = "The battery life is incredible and the camera quality blew me away, though the phone does run a bit warm during long gaming sessions."
```

```python
def summarize(text):
    response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": f"Summarize this review in one sentence:\n{text}"}])
    return response.choices[0].message.content
def classify(summary):
    response = client.chat.completions.create(model="gpt-4o-mini", messages=[{"role": "user", "content": f"Classify this as positive or negative. Respond with only one word:\n{summary}"}])
    return response.choices[0].message.content
```

```python
summary = summarize(review)
classification = classify(summary)
print(summary)
print(classification)
```
